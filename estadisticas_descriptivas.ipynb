```python
import numpy as np
import statistics as st

# pesos de las manzanas en gr
manzanas = [150, 180, 200, 170, 150, 190,200, 210, 150, 180]

#Estadistica descriptiva
media = np.mean(manzanas) # promedio
mediana = np.median(manzanas) # mediana
moda = st.mode(manzanas) # moda
desviacion = np.std(manzanas) # desviación (saber la variavilidad de los datos

#resultados
print(f"Analisis de la cosecha de manzanas")
print(f"Media: {media} gramos")
print(f"Mediana: {mediana} gramos")
print(f"Moda: {moda} gramos")
print(f"Desviacion: {desviacion:.2f} gramos") # sólo con dos decimales .2f
```

```python
import statistics as std
import numpy as np

zanahorias = [300, 280, 290, 310, 275, 290,
              295, 315, 290, 280, 310, 305]

media = np.mean(zanahorias)
mediana = np.median(zanahorias)
desviacion = np.std(zanahorias)

print("¡CONCURSO DE ZANAHORIAS!")
print("------------------------------")
print(f"Número de zanahorias", len(zanahorias))
print(f"Media: {media} gramos")
print(f"Mediana: {mediana} gramos")
print(f"Desviacion: {desviacion:.2f} gramos")
```

```python
import matplotlib.pyplot as plt
plt.hist(zanahorias)

plt.title("Distribución del peso de las zanahorias")
plt.xlabel("Peso (gramos)")
plt.ylabel("Frecuencia")

plt.show()
```
